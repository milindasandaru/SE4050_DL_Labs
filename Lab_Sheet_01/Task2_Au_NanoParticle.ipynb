```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Au_nanoParticle_dataset.csv")
```

```python
df.columns
```

```python
df.head(5)
```

```python
features = ["N_total", "N_bulk", "N_surface", "R_avg"]

new_df = df[features]
new_df.head(20)
```

```python
new_df.describe()
```

```python
plt.figure(figsize=(20,5))

for i, column in enumerate(new_df.columns):
    plt.subplot(1,4,i+1)
    sns.histplot(new_df[column], kde=True)
    plt.title(column)
    plt.xlabel(column)
    plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
sns.pairplot(new_df)
plt.show()
```

```python
g = sns.PairGrid(new_df)

# Upper triangle: bivariate histogram
g.map_upper(sns.histplot)

# Diagonal: histogram with KDE
g.map_diag(sns.histplot, kde=True)

# Lower triangle: bivariate KDE plot
g.map_lower(sns.kdeplot)

plt.show()
```
