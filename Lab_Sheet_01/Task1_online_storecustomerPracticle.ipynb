```python
import matplotlib.pyplot as plt
```

```python
uniform = np.random.rand(100000)
normal = np.random.normal(size=100000)
exponential = np.random.exponential(size=100000)
```

```python
plt.figure(figsize=(10,6))

plt.hist(uniform,
         density=True,
         bins=120,
         histtype='step',
         color='blue',
         label='Uniform')

plt.hist(normal,
         density=True,
         bins=120,
         histtype='step',
         color='green',
         label='Normal')

plt.hist(exponential,
         density=True,
         bins=120,
         histtype='step',
         color='red',
         label='Exponential')

# plt.axis([-4,8,0,1.2])
# plt.title("Comparison of Random Distributions")
# plt.xlabel("Value")
# plt.ylabel("Density")
# plt.legend()
# plt.show()

plt.axis([-2.5, 2.5, 0, 1.1]) 
plt.legend(loc = "upper left") 
plt.title("Random distributions") 
plt.xlabel("Value") 
plt.ylabel("Density") 
plt.show() 
```

```python
print("Uniform min:", uniform.min(), "max:", uniform.max())

print("Normal min:", normal.min(), "max:", normal.max())

print("Exponential min:", exponential.min(), "max:", exponential.max())
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

X,Y = np.meshgrid(x,y)

Z = X**2 + Y**2

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X,Y,Z,cmap='viridis')

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

ax.set_title("3D Plot of Z = X² + Y²")
plt.show()
```

```python
import pandas as pd
```

```python
customer = pd.read_csv("online_store_customer_data.csv")
```

```python
customer.head(5)
```

```python
customer.shape
customer.dtypes
customer.describe()
```

```python
import seaborn as sns
```

```python
features = ["Age", "Referal", "Amount_spent"]

pearson = customer[features].corr(method="pearson")
spearman = customer[features].corr(method="spearman")
```

```python
print(pearson)
```

```python
print(spearman)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    pearson,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Pearson Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    spearman,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Spearman Correlation Heatmap")

plt.show()
```
